## 2. Justificación de RAG + LLM

El problema que se busca resolver en este proyecto no es únicamente generar texto sobre inteligencia artificial, sino responder preguntas académicas del curso `IAyAA` con apego a los materiales reales de estudio. Un `LLM` usado de forma aislada puede producir respuestas fluidas y plausibles, pero también corre el riesgo de mezclar conceptos, omitir matices vistos en clase o responder de manera demasiado general. En un contexto académico esto es especialmente delicado, porque el estudiante necesita explicaciones consistentes con el contenido del curso y no solo definiciones amplias tomadas del conocimiento general del modelo.

La arquitectura `RAG` permite reducir ese problema al recuperar primero fragmentos relevantes del corpus documental y usar después esos fragmentos como contexto para la generación. En este caso, la recuperación se hace sobre transcripciones y documentos teóricos seleccionados del curso, lo que vuelve más probable que la respuesta incluya terminología, ejemplos y enfoques alineados con `IAyAA`. Así, el sistema no depende únicamente de la memoria paramétrica del modelo, sino de evidencia textual concreta extraída del dominio que se desea consultar.

La combinación `RAG + LLM` también se justifica porque equilibra precisión y claridad expositiva. El componente de recuperación aporta pertinencia temática y trazabilidad documental, mientras que el `LLM` transforma el contexto recuperado en una respuesta integrada, legible y útil para el usuario final. Esto resulta adecuado para un chatbot académico en español, ya que permite contestar preguntas conceptuales, comparativas y explicativas con base documental, sin requerir reentrenar el modelo ni construir una infraestructura compleja para la primera versión del proyecto.

## 3. Selección del LLM

Para esta primera versión del chatbot se eligió trabajar con OpenAI y usar `gpt-5-mini` como configuración base del componente generativo. La decisión no busca maximizar sofisticación, sino mantener un equilibrio adecuado entre disponibilidad, facilidad de integración y calidad suficiente para una entrega académica en notebook. Dado que el proyecto requiere responder en español con apoyo de contexto recuperado, interesa un modelo capaz de seguir instrucciones con precisión, sintetizar información técnica y operar de forma estable dentro de un flujo de prompting contextual.

El criterio de selección prioriza cuatro factores. Primero, la facilidad de integración en Python mediante el cliente oficial de OpenAI, lo que simplifica la conexión entre recuperación, construcción de prompt y generación final. Segundo, la disponibilidad práctica en el entorno de trabajo, ya que la integración puede activarse de forma controlada dentro del notebook y permite trabajar con credenciales sin exponerlas en el contenido visible del entregable. Tercero, el costo operativo, porque `gpt-5-mini` ofrece un perfil razonable para pruebas académicas y demostraciones iterativas sin exigir una configuración pesada. Cuarto, el desempeño esperado en comprensión y redacción en español, suficiente para transformar el contexto recuperado en respuestas claras para un estudiante de maestría.

También se mantuvo una decisión de diseño importante: desacoplar la interfaz generativa del resto del sistema. Aunque la configuración actual usa OpenAI con `gpt-5-mini`, el notebook conserva una estructura basada en una función tipo `generate_answer(contexto, pregunta)` para que el proveedor pueda sustituirse sin rehacer el pipeline RAG. Esta elección vuelve más robusto el entregable, porque permite justificar el modelo seleccionado hoy y, al mismo tiempo, conservar flexibilidad técnica si en otra ejecución se requiere cambiar de modelo, ajustar costos o trabajar temporalmente con un modo de respaldo extractivo.


**Alternativas consideradas (abiertas y propietarias):**

| Modelo | Tipo | Ventajas | Rol en este proyecto |
|---|---|---|---|
| `gpt-5-mini` (OpenAI) | Propietario | Excelente español, API estable, costo bajo por consulta, seguimiento fino de instrucciones | **Backend principal** (`openai`) |
| `Qwen2.5-7B-Instruct` (Alibaba) | Código abierto | Buen desempeño en español, licencia permisiva, disponible vía Hugging Face Inference API | Backend `hf`: alternativa abierta en la nube, usada en el despliegue del Space |
| `llama3.2:3b` (Meta) | Código abierto | Corre 100% local en CPU con 8 GB de RAM vía Ollama; sin costo por token ni dependencia de red | Backend `ollama`: alternativa abierta local |
| Respaldo extractivo | — | Sin dependencias externas, siempre disponible | Backend `extractivo`: degradación elegante |

A diferencia de una justificación puramente teórica, en este proyecto la comparación entre modelos
abiertos y propietarios está **implementada**: la sección 5.5 define los cuatro backends como funciones
intercambiables y la variable `LLM_BACKEND` (o el selector de la interfaz gráfica) permite elegir entre
ellos sin modificar el pipeline RAG. Esto demuestra empíricamente la decisión de diseño de desacoplar
la generación de la recuperación: el mismo contexto recuperado puede ser respondido por un modelo
propietario (`gpt-5-mini`), por un modelo abierto en la nube (`Qwen2.5-7B`) o por un modelo abierto
ejecutándose localmente (`llama3.2:3b`).

## 5. Implementación del sistema RAG

Esta sección concentra el desarrollo técnico del chatbot y describe cómo se articula el flujo completo desde la lectura del corpus hasta la generación de la respuesta final. La arquitectura implementada sigue un esquema `RAG` demostrable dentro del notebook, priorizando claridad, estabilidad y trazabilidad sobre complejidad de infraestructura.

### Arquitectura del sistema

El siguiente diagrama resume el flujo completo, desde la ingesta del corpus hasta la generación de la respuesta con fuentes citadas:

![Flujo del sistema RAG + LLM](capturas/diagrama/diagrama1.png)


El primer bloque del sistema corresponde a la **carga de documentos**. Para ello se leen archivos `.txt` y `.pdf` desde `04_datos_rag/` y se normalizan a una estructura común con campos como `source`, `text`, `tipo` y `tema`. Esta homogeneización permite tratar de forma consistente documentos provenientes de transcripciones y materiales teóricos, y además deja preparada la base para auditar qué fuente alimenta cada respuesta.

El segundo bloque es la **preparación del texto y el chunking**. Después de una limpieza ligera de espacios, saltos de línea y fragmentos ruidosos, cada documento se divide en bloques manejables (900 caracteres con traslape de 150). El propósito del chunking es conservar suficiente contexto local sin obligar al sistema a comparar preguntas contra documentos completos demasiado largos. Así, cada chunk se vuelve la unidad básica de recuperación semántica y de trazabilidad dentro del notebook.

El tercer bloque cubre la **representación vectorial y la recuperación semántica**. Los chunks se codifican con el modelo de embeddings neuronales multilingüe `paraphrase-multilingual-MiniLM-L12-v2` (`sentence-transformers`), que proyecta cada fragmento y cada consulta a un espacio vectorial denso donde la cercanía refleja similitud de significado y no solo coincidencia de palabras. La consulta del usuario se codifica en el mismo espacio y se rankean los fragmentos por similitud coseno para recuperar un conjunto `top-k`. Como evidencia del beneficio de este enfoque, se conserva el recuperador `TF-IDF` de la primera versión como *baseline* comparativo y la sección 8 cuantifica la diferencia entre ambos.

El cuarto bloque corresponde a la **construcción del prompt y la generación de respuesta**. Los fragmentos recuperados se formatean como contexto explícito y se incorporan a un prompt que obliga al modelo a responder en español, limitarse al material recuperado y declarar insuficiencia si la evidencia no basta. El componente generativo es **intercambiable**: `gpt-5-mini` (propietario, OpenAI), `claude-opus-4-8` (propietario, Anthropic), `Qwen2.5-7B-Instruct` (abierto, vía Hugging Face Inference API), `llama3.2:3b` (abierto, ejecución local con Ollama) o un modo extractivo de respaldo. De esta manera, el sistema conserva continuidad funcional incluso cuando ningún `LLM` está disponible.

Las **librerías principales** utilizadas en esta implementación responden a un criterio de simplicidad. `pandas` organiza documentos, chunks y resultados en estructuras tabulares fáciles de inspeccionar; `pypdf` extrae texto de los materiales teóricos en PDF; `sentence-transformers` aporta el modelo de embeddings semánticos multilingües; `scikit-learn` provee `TfidfVectorizer` y `cosine_similarity` para el baseline léxico; `openai` y `huggingface_hub` conectan con los modelos generativos propietario y abierto; `gradio` construye la interfaz de chat; e `IPython.display` y módulos estándar (`pathlib`, `re`, `os`, `unicodedata`) apoyan visualización, rutas y limpieza.

Finalmente, el sistema conserva **límites conocidos** que deben reconocerse en la revisión académica. La calidad de la recuperación depende del ruido presente en algunas transcripciones y de la cobertura del corpus seleccionado. La lectura de PDFs puede heredar problemas de formato y la generación con modelos externos depende de credenciales y cuota; por ello se mantienen el backend local abierto y el respaldo extractivo.

En conjunto, el flujo implementado puede resumirse en la siguiente secuencia:

1. Carga del corpus.
2. Limpieza y preparación del texto.
3. Chunking.
4. Embeddings semánticos del corpus.
5. Recuperación semántica (top-k por similitud coseno).
6. Construcción del prompt.
7. Generación de respuesta (backend intercambiable).
8. Presentación de fuentes recuperadas.


```python
# 5.2 Carga y extracción de documentos
# Lectura de archivos .txt y .pdf con una estructura homogénea.

def infer_topic(path: Path) -> str:
    return TOPIC_MAP.get(path.stem, "Tema por clasificar")


def read_txt(path: Path) -> str:
    return path.read_text(encoding="utf-8", errors="ignore")


def read_pdf(path: Path) -> str:
    reader = PdfReader(str(path))
    pages = [(page.extract_text() or "") for page in reader.pages]
    return "\n".join(pages)


def load_document(path: Path) -> dict:
    suffix = path.suffix.lower()
    if suffix == ".txt":
        text = read_txt(path)
    elif suffix == ".pdf":
        text = read_pdf(path)
    else:
        raise ValueError(f"Tipo de archivo no soportado: {path.name}")

    return {
        "source": path.name,
        "text": text,
        "tipo": suffix.lstrip("."),
        "tema": infer_topic(path),
    }


def load_corpus(data_dir: Path = DATA_DIR, allowed_suffixes=(".txt", ".pdf")) -> pd.DataFrame:
    documents = []
    for path in sorted(data_dir.iterdir()):
        if path.is_file() and path.suffix.lower() in allowed_suffixes:
            documents.append(load_document(path))
    return pd.DataFrame(documents)


documents_df = load_corpus()
documents_df["num_chars"] = documents_df["text"].str.len()
documents_df["num_words"] = documents_df["text"].str.split().str.len()

display(documents_df[["source", "tipo", "tema", "num_chars", "num_words"]])
print(f"Documentos cargados: {len(documents_df)}")
```

```python
# 5.3 Limpieza y chunking
# Preparar el corpus y fragmentarlo
# en bloques manejables para retrieval semántico.

def clean_text(text: str) -> str:
    text = text.replace("\r\n", "\n").replace("\r", "\n")
    text = re.sub(r"[\t\f\v]+", " ", text)
    text = re.sub(r" ?\n ?", "\n", text)
    text = re.sub(r"\n{3,}", "\n\n", text)
    text = re.sub(r" {2,}", " ", text)
    return text.strip()


def prepare_corpus(documents: pd.DataFrame, min_chars: int = 80) -> pd.DataFrame:
    prepared = documents.copy()
    prepared["text_raw"] = prepared["text"]
    prepared["text"] = prepared["text"].fillna("").map(clean_text)
    prepared["num_chars_clean"] = prepared["text"].str.len()
    prepared["num_words_clean"] = prepared["text"].str.split().str.len()
    prepared = prepared[prepared["num_chars_clean"] >= min_chars].reset_index(drop=True)
    return prepared


prepared_df = prepare_corpus(documents_df)

display(
    prepared_df[["source", "tipo", "tema", "num_chars", "num_chars_clean", "num_words_clean"]]
)

empty_docs = int((prepared_df["text"].str.strip() == "").sum())
print(f"Documentos listos para chunking: {len(prepared_df)}")
print(f"Documentos vacíos tras limpieza: {empty_docs}")

prepared_df[["source", "text"]].head(2)

CHUNK_SIZE = 900
CHUNK_OVERLAP = 150


def split_into_paragraphs(text: str) -> list[str]:
    paragraphs = [paragraph.strip() for paragraph in re.split(r"\n{2,}", text) if paragraph.strip()]
    return paragraphs or ([text.strip()] if text.strip() else [])


def split_long_paragraph(paragraph: str, chunk_size: int) -> list[str]:
    words = paragraph.split()
    windows = []
    current_words = []

    for word in words:
        candidate = " ".join(current_words + [word]).strip()
        if current_words and len(candidate) > chunk_size:
            windows.append(" ".join(current_words).strip())
            current_words = [word]
        else:
            current_words.append(word)

    if current_words:
        windows.append(" ".join(current_words).strip())

    return [window for window in windows if window]


def chunk_text(text: str, chunk_size: int = CHUNK_SIZE, chunk_overlap: int = CHUNK_OVERLAP) -> list[str]:
    paragraphs = split_into_paragraphs(text)
    chunks = []
    current_parts = []

    for paragraph in paragraphs:
        if len(paragraph) > chunk_size:
            if current_parts:
                chunks.append("\n\n".join(current_parts).strip())
                current_parts = []
            chunks.extend(split_long_paragraph(paragraph, chunk_size))
            continue

        candidate_parts = current_parts + [paragraph]
        candidate_text = "\n\n".join(candidate_parts).strip()

        if current_parts and len(candidate_text) > chunk_size:
            chunks.append("\n\n".join(current_parts).strip())
            if chunk_overlap > 0:
                overlap_text = chunks[-1][-chunk_overlap:].strip()
                current_parts = [overlap_text, paragraph] if overlap_text else [paragraph]
            else:
                current_parts = [paragraph]
        else:
            current_parts = candidate_parts

    if current_parts:
        chunks.append("\n\n".join(current_parts).strip())

    return [chunk for chunk in chunks if chunk]


def build_chunks(corpus: pd.DataFrame, chunk_size: int = CHUNK_SIZE, chunk_overlap: int = CHUNK_OVERLAP) -> pd.DataFrame:
    rows = []

    for _, row in corpus.iterrows():
        text_chunks = chunk_text(row["text"], chunk_size=chunk_size, chunk_overlap=chunk_overlap)
        for chunk_index, chunk in enumerate(text_chunks, start=1):
            rows.append(
                {
                    "chunk_id": f"{Path(row['source']).stem}_chunk_{chunk_index:03d}",
                    "source": row["source"],
                    "tipo": row["tipo"],
                    "tema": row["tema"],
                    "chunk_index": chunk_index,
                    "text": chunk,
                    "num_chars_chunk": len(chunk),
                    "num_words_chunk": len(chunk.split()),
                }
            )

    return pd.DataFrame(rows)


chunks_df = build_chunks(prepared_df)

display(chunks_df[["chunk_id", "source", "tema", "num_chars_chunk", "num_words_chunk"]].head(10))
display(chunks_df.groupby("source").size().reset_index(name="num_chunks"))
print(f"Chunks generados: {len(chunks_df)}")
print(f"Tamaño configurado por chunk: {CHUNK_SIZE} caracteres")
print(f"Traslape configurado: {CHUNK_OVERLAP} caracteres")
```

```python
# 5.4 Embeddings y recuperación semántica
# Recuperador principal: embeddings neuronales multilingües (sentence-transformers).
# La primera versión del sistema usaba TF-IDF con expansión manual de consultas;
# se conserva aquí como baseline comparativo (la sección 8 cuantifica la mejora).

from sentence_transformers import SentenceTransformer

EMBEDDING_MODEL_NAME = "paraphrase-multilingual-MiniLM-L12-v2"

st_model = SentenceTransformer(EMBEDDING_MODEL_NAME)
corpus_embeddings = st_model.encode(
    chunks_df["text"].tolist(),
    normalize_embeddings=True,
    show_progress_bar=True,
)

chunk_embeddings_df = chunks_df.copy()
chunk_embeddings_df["embedding_index"] = range(len(chunk_embeddings_df))
chunk_embeddings_df["embedding_model"] = EMBEDDING_MODEL_NAME

print(f"Modelo de embeddings activo: {EMBEDDING_MODEL_NAME}")
print(f"Matriz de embeddings del corpus: {corpus_embeddings.shape}")


def recuperar_contexto(pregunta: str, k: int = 5) -> pd.DataFrame:
    """Recuperación semántica: similitud coseno sobre embeddings normalizados."""
    query_embedding = st_model.encode([pregunta], normalize_embeddings=True)
    similarities = (corpus_embeddings @ query_embedding.T).flatten()
    top_indices = similarities.argsort()[::-1][:k]

    results = chunk_embeddings_df.iloc[top_indices].copy()
    results["score_similitud"] = similarities[top_indices]
    results["pregunta"] = pregunta
    return results[
        ["pregunta", "chunk_id", "source", "tema", "chunk_index", "score_similitud", "text"]
    ].reset_index(drop=True)


# --- Baseline TF-IDF (configuración de la primera versión, para comparación) ---

tfidf_vectorizer = TfidfVectorizer(
    lowercase=True,
    strip_accents="unicode",
    ngram_range=(1, 2),
    min_df=1,
    max_df=0.95,
    token_pattern=r"(?u)\b[a-záéíóúüñ]{2,}\b",
    sublinear_tf=True,
)
tfidf_embeddings = tfidf_vectorizer.fit_transform(chunks_df["text"])


def recuperar_contexto_tfidf(pregunta: str, k: int = 5) -> pd.DataFrame:
    """Baseline léxico: similitud coseno sobre vectores TF-IDF."""
    query_vec = tfidf_vectorizer.transform([pregunta])
    similarities = cosine_similarity(query_vec, tfidf_embeddings).flatten()
    top_indices = similarities.argsort()[::-1][:k]

    results = chunk_embeddings_df.iloc[top_indices].copy()
    results["score_similitud"] = similarities[top_indices]
    results["pregunta"] = pregunta
    return results[
        ["pregunta", "chunk_id", "source", "tema", "chunk_index", "score_similitud", "text"]
    ].reset_index(drop=True)


pregunta_prueba = "¿Qué es el gradiente descendente y para qué sirve en aprendizaje automático?"
print("\nTop-3 semántico:")
display(recuperar_contexto(pregunta_prueba, k=3)[["chunk_id", "source", "tema", "score_similitud"]])
print("Top-3 TF-IDF (baseline):")
display(recuperar_contexto_tfidf(pregunta_prueba, k=3)[["chunk_id", "source", "tema", "score_similitud"]])
```

```python
# 5.5 Prompt y componente generativo intercambiable
# Cuatro backends: OpenAI (propietario), Hugging Face y Ollama (código abierto)
# y un respaldo extractivo local. Sin credenciales expuestas en el notebook.

PROMPT_BASE = """Rol: Eres un asistente académico del curso Inteligencia Artificial y Aprendizaje Automático (IAyAA).

Objetivo:
- Responder preguntas del curso apoyándote en el contexto recuperado del material oficial.

Instrucciones obligatorias:
- Responde siempre en español.
- Basa tu respuesta en la información del contexto recuperado; puedes reorganizarla y sintetizarla con tus propias palabras.
- Si el contexto cubre el tema solo de forma parcial, responde con lo que sí está presente y aclara en una frase qué aspecto no quedó cubierto.
- Únicamente si el contexto no contiene nada relevante para la pregunta, inicia con: 'Con el contexto recuperado no es posible responder con suficiente precisión'.
- No inventes definiciones, fórmulas o ejemplos ajenos al contexto.
- Tono claro y preciso para un estudiante de maestría; menciona de forma breve la fuente o tema del que sale la respuesta.

Formato de salida esperado:
- Respuesta de 3 a 6 oraciones.
"""

LLM_BACKEND = os.environ.get("LLM_BACKEND", "auto")  # auto | openai | claude | hf | ollama | extractivo
OPENAI_MODEL = os.environ.get("OPENAI_MODEL", "gpt-5-mini")
OPENAI_REASONING_EFFORT = os.environ.get("OPENAI_REASONING_EFFORT", "low")
ANTHROPIC_MODEL = os.environ.get("ANTHROPIC_MODEL", "claude-opus-4-8")
HF_MODEL = os.environ.get("HF_MODEL", "Qwen/Qwen2.5-7B-Instruct")
OLLAMA_MODEL = os.environ.get("OLLAMA_MODEL", "llama3.2:3b")
OLLAMA_BASE_URL = os.environ.get("OLLAMA_BASE_URL", "http://localhost:11434/v1")


def format_contexto(contexto_df: pd.DataFrame) -> str:
    bloques = []
    for row in contexto_df.itertuples(index=False):
        bloques.append(
            f"[Fuente: {row.source} | Chunk: {row.chunk_id} | Score: {row.score_similitud:.4f}]\n{row.text}"
        )
    return "\n\n".join(bloques)


def construir_prompt(pregunta: str, contexto_df: pd.DataFrame) -> str:
    contexto = format_contexto(contexto_df)
    return (
        f"{PROMPT_BASE}\n"
        "Contexto recuperado:\n"
        "--------------------\n"
        f"{contexto}\n\n"
        "Pregunta del usuario:\n"
        "---------------------\n"
        f"{pregunta}\n\n"
        "Respuesta final:\n"
    )


def get_openai_llm_callable():
    """Backend propietario: gpt-5-mini vía API de OpenAI."""
    api_key = os.environ.get("OPENAI_API_KEY")
    if not api_key and IN_COLAB:
        from getpass import getpass
        api_key = getpass("OPENAI_API_KEY (Enter para omitir): ").strip()
        if api_key:
            os.environ["OPENAI_API_KEY"] = api_key
    if not api_key:
        return None

    from openai import OpenAI
    client = OpenAI(api_key=api_key)

    def llm_openai(prompt: str) -> str:
        response = client.responses.create(
            model=OPENAI_MODEL,
            input=prompt,
            reasoning={"effort": OPENAI_REASONING_EFFORT},
            store=False,
        )
        return response.output_text.strip()

    return llm_openai


def get_claude_llm_callable():
    """Backend propietario alternativo: Claude vía el SDK oficial de Anthropic."""
    api_key = os.environ.get("ANTHROPIC_API_KEY")
    if not api_key:
        return None

    import anthropic
    client = anthropic.Anthropic(api_key=api_key)

    def llm_claude(prompt: str) -> str:
        response = client.messages.create(
            model=ANTHROPIC_MODEL,
            max_tokens=1024,
            messages=[{"role": "user", "content": prompt}],
        )
        return next(
            (block.text for block in response.content if block.type == "text"), ""
        ).strip()

    return llm_claude


def get_hf_llm_callable():
    """Backend de código abierto en la nube: Qwen2.5-7B vía HF Inference API."""
    token = os.environ.get("HF_TOKEN")
    if not token:
        return None

    from huggingface_hub import InferenceClient
    client = InferenceClient(model=HF_MODEL, token=token)

    def llm_hf(prompt: str) -> str:
        response = client.chat_completion(
            messages=[{"role": "user", "content": prompt}],
            max_tokens=500,
            temperature=0.2,
        )
        return response.choices[0].message.content.strip()

    return llm_hf


def get_ollama_llm_callable():
    """Backend de código abierto local: llama3.2:3b servido por Ollama."""
    import urllib.request
    try:
        urllib.request.urlopen(OLLAMA_BASE_URL.removesuffix("/v1"), timeout=2)
    except Exception:
        return None

    from openai import OpenAI
    client = OpenAI(base_url=OLLAMA_BASE_URL, api_key="ollama")

    def llm_ollama(prompt: str) -> str:
        response = client.chat.completions.create(
            model=OLLAMA_MODEL,
            messages=[{"role": "user", "content": prompt}],
            temperature=0.2,
        )
        return response.choices[0].message.content.strip()

    return llm_ollama


_BACKEND_FACTORIES = {
    "openai": get_openai_llm_callable,
    "claude": get_claude_llm_callable,
    "hf": get_hf_llm_callable,
    "ollama": get_ollama_llm_callable,
}


def get_llm_callable(backend: str = "auto"):
    """Devuelve (callable, nombre). En modo auto prueba openai -> hf -> ollama."""
    backend = (backend or "auto").lower()
    if backend == "extractivo":
        return None, "extractivo"
    if backend in _BACKEND_FACTORIES:
        return _BACKEND_FACTORIES[backend](), backend
    for name in ("openai", "claude", "hf", "ollama"):
        llm = _BACKEND_FACTORIES[name]()
        if llm is not None:
            return llm, name
    return None, "extractivo"


def extractive_fallback(contexto_df: pd.DataFrame, max_sentences: int = 2) -> str:
    if contexto_df.empty:
        return "No encontré contexto suficiente en el corpus para responder con confianza."
    top_text = contexto_df.iloc[0]["text"]
    sentences = [s.strip() for s in re.split(r"(?<=[.!?])\s+|\n+", top_text) if s.strip()]
    selected = sentences[:max_sentences] if sentences else [top_text[:320].strip()]
    return f"Respuesta construida directamente a partir del fragmento recuperado: {' '.join(selected).strip()}"


def generar_respuesta(pregunta: str, contexto_df: pd.DataFrame, llm_callable=None) -> dict:
    prompt = construir_prompt(pregunta, contexto_df)
    if llm_callable is None:
        return {
            "respuesta": extractive_fallback(contexto_df),
            "prompt": prompt,
            "modo_generacion": "extractivo_local",
            "error_generacion": None,
        }
    try:
        return {
            "respuesta": llm_callable(prompt),
            "prompt": prompt,
            "modo_generacion": "llm",
            "error_generacion": None,
        }
    except Exception as exc:
        return {
            "respuesta": extractive_fallback(contexto_df),
            "prompt": prompt,
            "modo_generacion": "respaldo_por_error",
            "error_generacion": f"{type(exc).__name__}: {exc}",
        }


llm_por_defecto, BACKEND_ACTIVO = get_llm_callable(LLM_BACKEND)
print(f"Backend solicitado: {LLM_BACKEND} | Backend activo: {BACKEND_ACTIVO}")


def chatbot_rag(pregunta: str, k: int = 5, llm_callable=None) -> dict:
    contexto_df = recuperar_contexto(pregunta, k=k)
    generacion = generar_respuesta(pregunta, contexto_df, llm_callable=llm_callable)
    modo = BACKEND_ACTIVO if generacion["modo_generacion"] == "llm" else generacion["modo_generacion"]
    return {
        "pregunta": pregunta,
        "respuesta": generacion["respuesta"],
        "prompt": generacion["prompt"],
        "modo_generacion": modo,
        "error_generacion": generacion["error_generacion"],
        "contexto": contexto_df,
        "fuentes": contexto_df[["chunk_id", "source", "tema", "score_similitud"]].copy(),
    }


ejemplo_chatbot = chatbot_rag(
    "¿Qué es el gradiente descendente y para qué sirve en aprendizaje automático?",
    k=5,
    llm_callable=llm_por_defecto,
)
print(f"Modo generativo actual: {ejemplo_chatbot['modo_generacion']}")
if ejemplo_chatbot["error_generacion"]:
    print(f"Error al usar LLM: {ejemplo_chatbot['error_generacion']}")
print("Vista previa del prompt:\n")
print(ejemplo_chatbot["prompt"][:1200] + ("..." if len(ejemplo_chatbot["prompt"]) > 1200 else ""))
print()
print(ejemplo_chatbot["respuesta"])
display(ejemplo_chatbot["fuentes"])
```

```python
# 8.1 Evaluación cuantitativa del retrieval: hit rate de la fuente esperada
# Compara el recuperador semántico (principal) contra el baseline TF-IDF.

def evaluar_hit_rate(bateria: list[dict], retriever, k: int = DEMO_TOP_K) -> pd.DataFrame:
    filas = []
    for item in bateria:
        contexto = retriever(item["pregunta"], k=k)
        fuentes_recuperadas = set(contexto["source"])
        esperadas = {f.strip() for f in item["fuente_esperada"].split("+")}
        filas.append(
            {
                "id": item["id"],
                "tipo": item["tipo"],
                "hit@k (todas las esperadas)": esperadas.issubset(fuentes_recuperadas),
                "hit@1 (top-1 esperada)": contexto.iloc[0]["source"] in esperadas,
                "top1_source": contexto.iloc[0]["source"],
                "score_top1": round(float(contexto.iloc[0]["score_similitud"]), 4),
            }
        )
    return pd.DataFrame(filas)


print("Recuperador semántico (principal):")
hits_semantico = evaluar_hit_rate(bateria_preguntas, recuperar_contexto)
display(hits_semantico)
print(f"hit@{DEMO_TOP_K}: {hits_semantico['hit@k (todas las esperadas)'].mean():.0%} | "
      f"hit@1: {hits_semantico['hit@1 (top-1 esperada)'].mean():.0%}\n")

print("Baseline TF-IDF (primera versión):")
hits_tfidf = evaluar_hit_rate(bateria_preguntas, recuperar_contexto_tfidf)
display(hits_tfidf)
print(f"hit@{DEMO_TOP_K}: {hits_tfidf['hit@k (todas las esperadas)'].mean():.0%} | "
      f"hit@1: {hits_tfidf['hit@1 (top-1 esperada)'].mean():.0%}")
```
